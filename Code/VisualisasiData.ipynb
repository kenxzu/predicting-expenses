```python
dataset = pd.DataFrame(dataset)
```

```python
dataset
```

```python
dataset['Pengeluaran'] = dataset['Price']
```

```python
dataset.drop(columns='Price', inplace=True, axis=1)
```

```python
dataset
```

```python
from sklearn import linear_model
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

X = dataset[['ID']].copy()
y = dataset[['Pengeluaran']].copy()

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.20) 
  
# Splitting the data into training and testing data 
regr = LinearRegression() 
  
regr.fit(X_train, y_train) 
print(regr.score(X_test, y_test))

y_pred = regr.predict(X_test) 

# The coefficients
print("Coefficients: \n", regr.coef_)
# The mean squared error
print("Mean squared error: %.2f" % mean_squared_error(y_test, y_pred))
# The coefficient of determination: 1 is perfect prediction
print("Coefficient of determination: %.2f" % r2_score(y_test, y_pred))

# Plot outputs
plt.title('prediction')
plt.xlabel('Day')
plt.ylabel('Pengeluaran')
plt.scatter(X_test, y_test, color="green")
plt.scatter(X_train, y_train, color='black')
plt.plot(X_test, y_pred, color="blue", linewidth=3)

plt.xticks()
plt.yticks()

plt.show()
```

**mencari jam yang sering melakukan transaksi berdasarkan dataset awal**

```python
df2 = pd.read_csv("databelanjaku.csv")
```

```python
df2['Timestamp'] = pd.to_datetime(df2['Timestamp'], format ='%d/%m/%Y %H:%M')
```

```python
df2['Hour'] = df2['Timestamp'].dt.hour
```

```python
df2.head(20)
```

```python
mode_hour = df2['Hour'].mode()

print("Mode of 'Hour' column:", mode_hour)
```

```python

# Count occurrences of each hour
hour_counts = df2['Hour'].value_counts()
hour_counts
```

```python


# Plotting
plt.figure(figsize=(8, 8))
plt.pie(hour_counts.values, labels=hour_counts.index, autopct='%1.1f%%', startangle=140)
plt.title('Distribution of Hours')
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```

simpulan : jam 3 sore, 8 malam, 1 malam, 11 siang, 6 sore adalah paling banyak dari melakukan transaksi di 40 hari terakhir.
