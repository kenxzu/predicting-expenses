```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"
column_names = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'class']


iris_data = pd.read_csv(url, header=None, names=column_names)


iris_data
```

```python
X = iris_data.drop(columns=['class'])
y = iris_data['class']  

#Data dibagi dengan proporsi : a. Data Latih (70%), dan b. Data Uji (30%)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)
```

**K-fold Cross val**


```python
from sklearn.model_selection import cross_val_score

from sklearn.svm import SVC
model = SVC(kernel='linear', C = 1)
scores = cross_val_score(model, X, y, cv = 3)
scores.mean()

#ini setelah saya melakukan hyperparametic tuning dengan yang awalnya mengggunakan C = 1 dan cv = 5 hanya  mendapatkan akurasi 
#skor 96% setelah melelakukan beberapa percobaan dalam hyperparameter tunning saya mendapatkan score akurasi yang lebih baik 
#dengan kernel = linear  C = 1 dan cross validataion sebesar 3 dengan akurasi skor 99%.
```

```python
from sklearn.model_selection import cross_val_score
```

**K-nearest neighbors**


```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn import metrics
#nilai default n_neighbors =3 ; sakurasi skor = 0.95555 atau setara dengan 95%
# namun dengan dengan n_neigbors = 7 score acurasi = 100% yang berarti tunning yang dilakukan cocok untuk data ini.
knn= KNeighborsClassifier(n_neighbors = 7)
knn.fit(X_train, y_train)
y_pred = knn.predict(X_test)
score  = metrics.accuracy_score(y_test, y_pred)


score
```

**Decision Tree**

```python
from sklearn.tree import DecisionTreeClassifier

#setelah dilakukan tunning skor terbaik adalah dengan parameter max_depth = 3, jumlah split 2
#max def 3 = 0.97, 

dtree = DecisionTreeClassifier(max_depth=3, min_samples_split=2)

dtree.fit(X_train, y_train)

y_pred = dtree.predict(X_test)
score = metrics.accuracy_score(y_test, y_pred)
score
```

**Naive Bayes**


```python
#data tidak cocok dengan model naive_bayes 
from sklearn import naive_bayes
nb= naive_bayes.BernoulliNB()
nb.fit(X_train, y_train)
y_pred = nb.predict(X_test)
scorenb = metrics.accuracy_score(y_test, y_pred)


scorenb
```

**SVM**


```python
from sklearn.svm import SVC

#disini saya melakukan paremeter tuning pada kernel svc yang pertama adalah rbf = 91%, poly = 86%, sigmoid = 66%, 
#namun ketika menggunakan kerner linear saya mendapatkan akurasi 100% 


model = SVC(kernel='linear', C = 1, gamma=0.01)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
scoresvm = metrics.accuracy_score(y_test, y_pred)
scoresvm

```

**AdaBoost**


```python
from sklearn.ensemble import AdaBoostClassifier
from sklearn import metrics
#setelah berebera kali melakukan tunning parameters (n_estimator) tidak ada perubahan yang signifikan pada nilai skor.
#nilai skornya hanya mencapai 93% 
#berbeda dengan parameter learning rate saya  menemukan ada perubhan skor yang terjadi terutama  pada learning rate 21
#ada peningkatan nilai skor yang awalnya 93% menjadi 97%
#yang dilakukan learning rate tunning parameters 
model = AdaBoostClassifier(
    n_estimators=50, learning_rate=21
)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
scoreadabost =metrics.accuracy_score(y_test, y_pred)
scoreadabost

```

Setelah melakukan hyper parameter tunning terhadap model diata
model K-Fold Cross validation menghasilkan akurasi skor 99%, 
model K-nearest neighbors menghasilkan akurasi skor 100%,
 model Decision Tree menghasilkan akurasi skor 97%,
 model Naive Bayes hanya menghasilkan akurasi skor sebesar 28%,
 model SVM menghasilkan akurasi skor 100%,
 model AdaBoost menghasilkan akurasi skor 97%

 jadi, model yang paling cocok untuk data iris ini adalah Model SVM dan K-nearest neighbors. 
 Disusul dengan model k-fold yang skor akurasinya sedikit terpaut dengan kedua model terbaik.
Kemudian model yang kurang cocok dengan data adalah Decision Tree dan AdaBoost dimana keduanya mempunyai akurasi yang sama. 
Dan Model yang sangat tidak cocok dengan data ini adalah model classifikasi Naive Bayes yang hanya menghasilkan 28%.
