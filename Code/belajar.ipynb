```python
#membuat bar/histogram dari data 
import numpy as np
p = df.index
q = df['Attack Speed Total']
warna = ['#0000FF' for _ in range(len(df))]


df.sort_values(by='Attack Speed Total',  ascending=False, inplace=True)

x_coords = np.arange(len(df))

plt.figure(figsize=(20,20))
plt.bar(x_coords,q, tick_label=p, color = warna, width=1, edgecolor ='black')
plt.xticks(x_coords, p, rotation = 90)
plt.xlabel('Hero')
plt.ylabel('attack speed')
plt.grid(True)
plt.show()
```

```python
from sklearn.feature_selection import SelectKBest
from sklearn.feature_selection import chi2
import numpy as np 
import pandas as pd


X =df.iloc[:,0:11]
y = df.iloc[:,-1].astype(int)#this method the target should be the in type so it works
bestfeatures = SelectKBest(score_func=chi2, k=5)
```

```python
y
```

```python
X.dtypes
```

```python
fit = bestfeatures.fit(X, y)
fit_data = fit.scores_
mytble = pd.DataFrame({'feature' : X.columns,'score' : fit_data})

mytble
```
